```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data Cleaning and Validation

```python
# Read the CSV file
file_path = 'heart_disease_health_indicators_BRFSS2015 2 (1).csv'
data = pd.read_csv(file_path)

# Print the first few lines of the CSV file
print(data.head())
```

```python


# Convert categorical variables to appropriate data types
# Assuming 'Sex' and 'Age' are categorical variables
data['Sex'] = data['Sex'].astype('category')
data['Age'] = data['Age'].astype('category')

# Perform a final check to ensure the dataset is clean and ready for analysis
# Check for missing values
missing_values = data.isnull().sum()

# Check data types
data_types = data.dtypes

# Print the results
print("Missing values in each column:\n", missing_values)
print("\nData types of each column:\n", data_types)

# Display the first few rows of the cleaned dataset
print("\nFirst few rows of the cleaned dataset:\n", data.head())
```

# Univariate Analysis

```python
# Set up the plotting style
sns.set(style="whitegrid")
```

```python
#a. Examine Distributions of Individual Variables:
# Plot histogram for Age
plt.figure(figsize=(10, 6))
sns.countplot(x='Age', data=data, palette='viridis')
plt.title('Distribution of Age')
plt.xlabel('Age')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
# Plot histogram for BMI
plt.figure(figsize=(10, 6))
sns.histplot(data['BMI'], bins=30, kde=True, color='blue')
plt.title('Distribution of BMI')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot histogram for Mental Health  (MentHlth)
plt.figure(figsize=(10, 6))
sns.histplot(data['MentHlth'], bins=30, kde=True, color='green')
plt.title('Distribution of Mental Health Days (MentHlth)')
plt.xlabel('Mental Health Days')
plt.ylabel('Frequency')
plt.show()
```

```python
# Plot histogram for Physical Health(PhysHlth)
plt.figure(figsize=(10, 6))
sns.histplot(data['PhysHlth'], bins=30, kde=True, color='red')
plt.title('Distribution of Physical Health Days (PhysHlth)')
plt.xlabel('Physical Health Days')
plt.ylabel('Frequency')
plt.show()
```

```python
# b. Investigate Prevalence of Health Conditions:

# Plot bar chart for High Blood Pressure (HighBP)
plt.figure(figsize=(10, 6))
sns.countplot(x='HighBP', data=data, palette='viridis')
plt.title('Prevalence of High Blood Pressure (HighBP)')
plt.xlabel('High Blood Pressure')
plt.ylabel('Count')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

```python
# Plot bar chart for High Cholesterol (HighChol)
plt.figure(figsize=(10, 6))
sns.countplot(x='HighChol', data=data, palette='viridis')
plt.title('Prevalence of High Cholesterol (HighChol)')
plt.xlabel('High Cholesterol')
plt.ylabel('Count')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

```python
# c. Analyze Distribution of Heart Disease (Target Variable):

# Plot bar chart for Heart Disease or Attack (HeartDiseaseorAttack)
plt.figure(figsize=(10, 6))
sns.countplot(x='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('Distribution of Heart Disease or Attack')
plt.xlabel('Heart Disease or Attack')
plt.ylabel('Count')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```

# Bivariate Analysis

```python
# Convert relevant columns to category
data['HeartDiseaseorAttack'] = data['HeartDiseaseorAttack'].astype('category')
data['HighBP'] = data['HighBP'].astype('category')
data['HighChol'] = data['HighChol'].astype('category')
data['Sex'] = data['Sex'].astype('category')
data['Age'] = data['Age'].astype('category')
data['Education'] = data['Education'].astype('category')
```

```python
# Ensure HeartDiseaseorAttack column is converted to strings
data['HeartDiseaseorAttack'] = data['HeartDiseaseorAttack'].astype(str)
data['HighBP'] = data['HighBP'].astype(str)
data['HighChol'] = data['HighChol'].astype(str)
data['Sex'] = data['Sex'].astype(str)
data['Age'] = data['Age'].astype(str)
data['Education'] = data['Education'].astype(str)
```

```python
# a. Explore Relationships with Heart Disease:

# Plot distribution of HighBP between individuals with and without heart disease
plt.figure(figsize=(10, 6))
sns.countplot(x='HighBP', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('High Blood Pressure by Heart Disease Status')
plt.xlabel('High Blood Pressure')
plt.ylabel('Count')
plt.xticks([0, 1], ['No', 'Yes'])
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```

```python
# Plot distribution of HighChol between individuals with and without heart disease
plt.figure(figsize=(10, 6))
sns.countplot(x='HighChol', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('High Cholesterol by Heart Disease Status')
plt.xlabel('High Cholesterol')
plt.ylabel('Count')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```

```python
# Plot distribution of BMI between individuals with and without heart disease
plt.figure(figsize=(10, 6))
sns.boxplot(x='HeartDiseaseorAttack', y='BMI', data=data, palette='viridis')
plt.title('BMI by Heart Disease Status')
plt.xlabel('Heart Disease or Attack')
plt.ylabel('BMI')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```

```python
# b. Visualize Correlations Between Variables:

# Compute the correlation matrix
corr_matrix = data.corr(numeric_only=True)
```

```python
# Plot the heatmap
plt.figure(figsize=(12, 10))
sns.heatmap(corr_matrix, annot=True, cmap='viridis', linewidths=0.5)
plt.title('Correlation Matrix')
plt.show()
```

```python
# Scatter plot for BMI and Age
plt.figure(figsize=(10, 6))
sns.scatterplot(x='Age', y='BMI', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('Scatter Plot of BMI vs Age')
plt.xlabel('Age')
plt.ylabel('BMI')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```

```python
# c. Compare Heart Disease Across Demographic Groups:

# Plot prevalence of heart disease across different age groups
plt.figure(figsize=(14, 8))
sns.countplot(x='Age', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('Heart Disease Prevalence Across Age Groups')
plt.xlabel('Age')
plt.ylabel('Count')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.xticks(rotation=45)
plt.show()
```

```python
# Plot prevalence of heart disease across different genders
plt.figure(figsize=(10, 6))
sns.countplot(x='Sex', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('Heart Disease Prevalence by Gender')
plt.xlabel('Gender')
plt.ylabel('Count')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```

```python
# Plot prevalence of heart disease across different education levels
plt.figure(figsize=(14, 8))
sns.countplot(x='Education', hue='HeartDiseaseorAttack', data=data, palette='viridis')
plt.title('Heart Disease Prevalence Across Education Levels')
plt.xlabel('Education Level')
plt.ylabel('Count')
plt.legend(title='Heart Disease or Attack', labels=['No', 'Yes'])
plt.show()
```
